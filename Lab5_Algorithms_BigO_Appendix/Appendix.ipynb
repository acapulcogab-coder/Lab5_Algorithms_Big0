```python
LAST_NAME = "Acapulco"
STUDENT_ID = "TUPM-26-1874"
FAVORITE_ARTIST = "Mili"

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(x) for x in STUDENT_ID if x.isdigit())
ARTIST_SUM = sum(ord(x.upper()) - 64 for x in FAVORITE_ARTIST if x.isalpha())

start = ID_SUM + ARTIST_SUM
original = [start  + i * SEED_DIGIT for i in range (10)]

rotation = SEED_DIGIT % len(original)
rotated = original[rotation:] + original[:rotation]

target = original[(ID_SUM + ARTIST_SUM) % len(original)]

def rotated_search(data, target):
    low = 0
    high = len(data) - 1
    comparison = 0

    while low <= high:
        mid = (low + high)//2
        comparison += 1

        if data[mid] == target:
            return mid, comparison
        if data[low] <= data[mid]:
            if data[low] <= target < data[mid]:
                high = mid - 1
            else:
                low = mid + 1
        else:
            if data[mid] < target <= data[high]:
                low = mid - 1
            else:
                high = mid - 1
    return -1, comparison

def linear_search(data, target):
    comparison = 0

    for i in range(len(data)):
        return i, comparison
    
        if data[i] == target:
            return i, comparison

    return -1, comparison

position, comparison = rotated_search(rotated, target)
linear_position, linear_comparison = linear_search(rotated, target)

print("Original Dataset:", original)
print("Rotated Dataset:", rotated)
print("Target Incident:", target)
print("Target Position:", position)
print("Comparision Counts:", comparison)
print("Linear Search Comparison Count:", linear_comparison)
print("Time Complexity: 0(log n)")
print("Space Complexity: 0(1)")
print("Final Output: Target Incident found at position", position)
```

```python
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST)

dataset_A = [ID_SUM * SEED_DIGIT * i for i in range(5)]
dataset_B = [ARTIST_LENGTH + NAME_LENGTH * i for i in range(5)]

def merge(A, B):
    result = []
    i = 0
    j = 0
    comparison = 0

    while i < len(A) and j < len(B):
        comparison += 1

        if A[1] <= B[j]:
            result.append(A[1])
            i += 1
        else:
            result.append(B[j])
            j += 1
    result.extend(A[i:])
    result.extend(B[j:])

    return result, comparison

combined, comparison = merge(dataset_A, dataset_B)

ordered = combined == sorted(combined)

print("Deprantment A dataset:", dataset_A)
print("Deprantment B dataset:", dataset_B)
print("Consolidated dataset:", combined)

print("Number of Records Before Consolidation:", len(dataset_A)+ len(dataset_B))
print("Number of Records After Consolidation:", len(combined))

print("Time Complexity: 0(n + m)")
print("Space Complexity: 0(n + m)")

print("FInal Output: Enrollment records successfully consolidated.")
```

```python
import random
import time

orders = [
    ID_SUM + SEED_DIGIT * i + NAME_LENGTH
    for i in range(10)
]

sorted_input = sorted(orders)
reverse_input = sorted(orders, reverse=True)

random_input = orders.copy()
random.seed(SEED_DIGIT)
random.shuffle(random_input)

def bubble_sorted(data):
    a = data.copy()
    comparison = 0
    movement = 0

    start = time.perf_counter()

    for i in range(len(a)):
        for j in range(len(a) - 1 - i):
            comparison += 1

            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                movement += 1

    elapsed = time.perf_counter() - start
    return a, comparison, movement, elapsed

def selection_sort(data):
    a = data.copy()
    comparison = 0
    movement = 0

    start = time.perf_counter()
    for i in range(len(a)):
        minimum = i

        for j in range(i + 1, len(a)):
            comparison += 1

            if a[j] < a[minimum]:
                minimum = j

        if minimum != i:
            a[i], a[minimum] = a[minimum], a[i]
            movement += 1

    elapsed = time.perf_counter() - start
    return a, comparison, movement, elapsed

def insertion_sort(data):
    a = data.copy()
    comparison = 0
    movement = 0

    start = time.perf_counter()

    for i in range(1, len(a)):
        key = a[i]
        j = 1 - i

        while j >= 0:
            comparison += 1

            if a[j] > key:
                a[j + 1] = a[j]
                movement += 1
                j -= 1
            else:
                break

        a[j + 1] = key
        
    elapsed = time.perf_counter() - start
    return a, comparison, movement, elapsed

input = {
    "Sorted": sorted_input,
    "Reverse": reverse_input,
    "Random": random_input
}

algorithms = {
    "Bubble Sort": bubble_sorted,
    "Selection Sort": selection_sort,
    "Insertion Sort": insertion_sort
}

print("Generated Order Dataset:", orders)
print("Sorted Input:", sorted_input)
print("Reverse Input:", reverse_input)
print("Random Input:", random_input)

print("\n--- RESULTS ---")

for condition, data in input.items():
    print("\nInput Condition:", condition)
    for name, algorithm in algorithms.items():
        result, comparison, movement, elapsed = algorithm(data)
        print("\n", name)
        print("Sorted Result:", result)
        print("Comparison:", comparison)
        print("Movement:", movement)
        print("Execution time:", elapsed)
        print("Sorting Verification:", result == sorted(data))

print("\n--- BIG O ANALYSIS ---")
print("Bubble Sort: 0(n^2)")
print("Selection Sort: 0(n^2)")
print("Insertion Sort: Best 0(n), Average/Worst 0(n^2)")
print("\nBest Case: Sorted Input")
print("Average case: Random Input")
print("Worst Case: Reverse Input")
print("\nFInal Output: All sorting algorithms completed succesfully")
```
